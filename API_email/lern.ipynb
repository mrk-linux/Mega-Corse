```python
import requests
from send_mail import send_email  # Import email sending function from send_mail.py


api_key = "your api key"  # API key to access the news service
url = "The URL of the news you want to read"  # News API URL you want to read from

# Send request to the API
request = requests.get(url)

# Convert the response to a dictionary (JSON)
content = request.json()

# Access article titles and descriptions
for article in content["articles"][:20]:  # Only process the first 20 articles
    if article["title"] is not None:  # Skip if the article title is empty
        body = "subject: Today's news" + "/n" + body + article["title"] + "/n" + article["description"] + 2 * "/n"

# Encode the message to UTF-8 format for email sending
body = body.encode("utf-8")

# Send the email
send_email(body)
```
