# Day 1 

### Add OpenCV script to capture and show webcam feed with 'q' exit key

```python
import cv2              # Import OpenCV library for image processing
import time             # Import time library for creating delays

video = cv2.VideoCapture(0)   # Initialize webcam (0 = default camera)
time.sleep(1)                 # One second pause to let the camera warm up

while True:                   # Start infinite loop to capture video frames
    check, frame = video.read()   # Read a frame from the camera (success flag / frame data)
    cv2.imshow("my video", frame) # Display the frame in a window titled "my video"

    key = cv2.waitKey(1)          # Wait 1 millisecond for a keyboard input

    if key == ord("q"):           # If the pressed key is 'q'
        break                     # Exit the infinite loop

video.release()                   # Release the camera and end webcam access
```

### Motion Detection with Bounding Boxes Using OpenCV

```python
import cv2              # Import OpenCV library for image processing
import time             # Import time library for creating delays

video = cv2.VideoCapture(0)   # Initialize webcam (0 = default camera)
time.sleep(1)                 # One second pause to let the camera warm up

first_frame = None            # Variable to store the first frame as background

while True:                   # Start infinite loop to capture video frames
    check, frame = video.read()   # Read a frame from the camera (success flag / frame data)
    
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    # Convert frame to grayscale for easier processing
    gray_frame_gau = cv2.GaussianBlur(gray_frame, (21, 21), 0)
    # Apply Gaussian blur to reduce noise

    if first_frame is None:       # Check if this is the first frame
        first_frame = gray_frame_gau   
        # Store the first frame as the reference background
        continue                  # Skip the rest of the loop for the first frame

    delta_frame = cv2.absdiff(first_frame, gray_frame_gau)  
    # Find absolute difference between first frame and current frame
    thresh_frame = cv2.threshold(delta_frame, 60, 255, cv2.THRESH_BINARY)[1]  
    # Apply threshold to highlight motion areas
    dil_frame = cv2.dilate(thresh_frame, None, iterations=2)  
    # Dilate to fill gaps in motion areas
    
    cv2.imshow("Motion Detection", dil_frame)  
    # Show the motion mask (white = motion detected)

    contours, check = cv2.findContours(dil_frame, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)  
    # Find contours of moving objects

    for contour in contours:      # Loop through all detected contours
        if cv2.contourArea(contour) < 5000:  # Ignore small contours (noise)
            continue              # Skip this contour and continue to next
    
        x, y, w, h = cv2.boundingRect(contour)  
        # Get bounding rectangle coordinates
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 3)  
        # Draw green rectangle around detected motion

    cv2.imshow("my video", frame)  # Display the original frame with motion rectangles
    
    key = cv2.waitKey(1)          # Wait 1 millisecond for a keyboard input

    if key == ord("q"):           # If the pressed key is 'q'
        break                     # Exit the infinite loop

video.release()                   # Release the camera and end webcam access
cv2.destroyAllWindows()           # Close all OpenCV windows (optional but recommended)
```

```python
import cv2              # Image processing library
import time             # Time library for delay

video = cv2.VideoCapture(0)   # Initialize webcam
time.sleep(1)                 # Delay for camera warm-up

first_frame = None            # Store first frame as background

while True:
    check, frame = video.read()   # Read frame from camera
    
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)  
    # Convert to grayscale
    gray_blur = cv2.GaussianBlur(gray, (21, 21), 0)  
    # Apply blur to reduce noise

    if first_frame is None:        # If this is the first frame
        first_frame = gray_blur    # Save as background
        continue                   # Skip to next frame

    delta = cv2.absdiff(first_frame, gray_blur)  
    # Find difference from background (motion detection)
    thresh = cv2.threshold(delta, 50, 255, cv2.THRESH_BINARY)[1]  
    # Apply threshold
    dilate = cv2.dilate(thresh, None, iterations=2)  
    # Fill gaps in motion areas

    contours, _ = cv2.findContours(dilate, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)  
    # Find objects

    object_number = 1   # Object counter
    
    for contour in contours:   # For each detected object
        if cv2.contourArea(contour) < 3000:  # If area is too small
            continue            # Skip (it's noise)
        
        x, y, w, h = cv2.boundingRect(contour)  
        # Get bounding rectangle coordinates
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)  
        # Draw green rectangle
        

        cv2.putText(frame, f"Object {object_number}", (x, y-10),  
                    # Display object number
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
        
        object_number += 1   # Increment counter

    cv2.imshow("Motion Detection", frame)  # Display video

    if cv2.waitKey(1) == ord("q"):  # If 'q' is pressed
        break                       # Exit loop

video.release()        # Release camera
cv2.destroyAllWindows()  # Close windows
```

### Add image saving feature to motion detection

```python
import cv2
import time
import glob  # For working with image files and listing them
from Email_ing import send_email


video = cv2.VideoCapture(0)
time.sleep(1)

first_frame = None
status_list = []
count = 1  # Counter for naming saved images

while True:
    status = 0
    check, frame = video.read()

    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray_frame_gau = cv2.GaussianBlur(gray_frame, (21, 21), 0)

    if first_frame is None:
        first_frame = gray_frame_gau

    delta_frame = cv2.absdiff(first_frame, gray_frame_gau)
    thresh_frame = cv2.threshold(delta_frame, 60, 255, cv2.THRESH_BINARY)[1]
    dil_frame = cv2.dilate(thresh_frame, None, iterations=2)
    cv2.imshow("My Video", dil_frame)

    contours, check = cv2.findContours(dil_frame, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    for contour in contours:
        if cv2.contourArea(contour) < 3000:
            continue
        
        x, y, w, h = cv2.boundingRect(contour)  
        rectangle = cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)  

        if rectangle.any():
            status = 1
            cv2.imwrite(f"images/{count}.png", frame)  # Save image to images folder
            count = count + 1  # Increment counter
            all_images = glob.glob("images/*.png")  # Get list of all saved images
            index = int(len(all_images) / 2)  # Select the middle image
            image_with_object = all_images[index]  # Middle image

    status_list.append(status)
    status_list = status_list[-2:]

    if status_list[0] == 1 and status_list[1] == 0:
        send_email(image_with_object)

        cv2.putText(frame, f"Object {object_number}", (x, y-10),  
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
        
        object_number += 1

    cv2.imshow("Motion Detection", frame)

    if cv2.waitKey(1) == ord("q"):
        break

video.release()
cv2.destroyAllWindows()
```

### Send Email

```python
import smtplib  # For connecting to email server and sending messages
import imghdr  # For detecting image type (e.g., png, jpg)
from email.message import EmailMessage  # For creating email with attachments

# Gmail login credentials
PASSWORD = "your_password_here"  # Gmail app password (not your regular password)
SENDER = "app8flask@gmail.com"  # Sender email address
RECEIVER = "app8flask@gmail.com"  # Receiver email address

def send_email(image_path):  # Function to send email with image
    email_message = EmailMessage()  # Create a new email message
    email_message["Subject"] = "New customer showed up!"  # Email subject
    email_message.set_content("Hey, we just saw a new customer!")  # Email body text

    with open(image_path, "rb") as file:  # Open image file in binary mode
        content = file.read()  # Read image content
        # Add image as attachment
        email_message.add_attachment(
            content,  # Image data
            maintype="image",  # Main type = image
            subtype=imghdr.what(None, content),  # Detect image type (png, jpg, etc.)
            filename=image_path.split("/")[-1]  # File name (without path)
        )

    # Connect to Gmail server
    gmail = smtplib.SMTP("smtp.gmail.com", 587)  # Gmail server and port
    gmail.ehlo()  # Say hello to server
    gmail.starttls()  # Enable security (TLS)
    gmail.login(SENDER, PASSWORD)  # Login to Gmail account
    gmail.sendmail(SENDER, RECEIVER, email_message.as_string())  # Send email
    gmail.quit()  # Disconnect
    print("✅ Email sent successfully!")  # Success message

if __name__ == "__main__":  # If this file is run directly
    send_email(image_path="images/19.png")  # Send email with specified image
```
