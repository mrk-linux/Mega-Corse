# Day 1

## Loading the book

```python
with open("miracle-in-the-andes.txt", "r", encoding="utf-8") as file:
    book = file.read()
```

## The most used words (non-articles)

```python
import re 


pattern = re.compile("[a-zA-Z]+")
findings = re.findall(pattern, book.lower())
findings[:5]
```

```python
d = {}
for word in findings:
    if word in d.keys():
        d[word] = d[word] + 1
    else:
        d[word] = 1
```

```python
d_list = [(value, key) for (key, value) in d.items()]
d_list = sorted(d_list, reverse=True)
d_list[:5]
```

```python
import nltk
from nltk.corpus import stopwords

english_stopwords = stopwords.words("english")
```

```python
filter_words = []
for count, word in d_list:
    if word not in english_stopwords:
        filter_words.append((word, count))
```

```python
filter_words[:10]
```

## sentiment Analysis: What is the most positive and the most negative chapter?

### An example

```python
from nltk.sentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()
```

```python
scores = analyzer.polarity_scores("Hey, look how beautiful the trees are. I Love them")
```

```python
analyzer.polarity_scores("Hey, look how bad the trees are. I hate them")
```

```python
if scores["pos"] > scores["neg"]:
    print("It is a positive text")
else:
    print("It is a negative text")
```

```python
analyzer.polarity_scores(book)
```

### Chapters sentiment analysis

```python
import re
pattern = re.compile("Chapter [0-9]+")
chapters = re. split(pattern, book)
```

```python
chapters = chapters[1:]
```

```python
for nr, chapter in enumerate(chapters):
    scores = analyzer.polarity_scores(chapter)
    print(nr+1, scores)
```
