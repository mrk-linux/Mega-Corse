# Day 1

Implement regex to count chapters in book file

## Load the book

```python
with open("miracle-in-the-andes.txt", "r", encoding="utf-8") as file:
    book = file.read()
```

## How many chapters?

### With string methods

```python
book.count("Chapter")
```

### with regex

```python
import re
pattern = re.compile("Chapter [0-9]+")
findings = re.findall(pattern, book)
len(findings)
```
