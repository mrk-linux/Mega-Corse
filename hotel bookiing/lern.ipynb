```python
import pandas  # Import pandas library for tabular data handling

# Read CSV file and set 'id' column as string type
df = pandas.read_csv("hotels.csv", dtype={"id": str})


class Hotel:  # Main Hotel class

    def __init__(self, hotel_id):  # Class constructor, receives hotel ID
        self.hotel_id = hotel_id

    def book(self):  # Method to book a hotel
        """Book a hotel changing its availability to NO"""
        df.loc[df["id"] == self.hotel_id, "available"] = "no"  # Change status to no
        df.to_csv("hotels.csv", index=False)  # Save changes to file

    def available(self):  # Method to check hotel availability
        """Check if hotel is available"""
        availability = df.loc[df["id"] == self.hotel_id, "available"].iloc[0]  # Get status
        if availability == "yes":
            return True
        else:
            return False


class ReservationTicket:  # Reservation ticket class

    def __init__(self, customer_name, hotel_object):  # Class constructor
        pass

    def generate(self):  # Method to generate ticket
        pass


print(df)

hotel_ID = input("Enter the id of the hotel:")  # Get hotel ID from user
hotel = Hotel(hotel_ID)

if hotel.available():  # If hotel is available
    hotel.book()  # Book the hotel
    name = input("Enter your name:")  # Get customer name
    rv = ReservationTicket(name, hotel)  # Create reservation ticket
    print(rv.generate())  # Print the ticket
else:
    print("Hotel is not free.")
```

```python
import pandas 


df = pandas.read_csv("hotels.csv", dtype={"id": str})


class Hotel:

    def __init__(self, hotel_id):
        self.hotel_id = hotel_id
        self.name = df.loc[df["id"] == self.hotel_id, "name"].squeeze()

    def book(self):
        """Book a hotel changing its availability to NO"""
        df.loc[df["id"] == self.hotel_id, "available"] = "no"
        df.to_csv("hotels.csv", index=False)

    def available(self):
        """Check if hotel is available"""
        availability = df.loc[df["id"] == self.hotel_id, "available"].iloc[0] 
        if availability == "yes":
            return True
        else:
            return False


class ReservationTicket:

    def __init__(self, customer_name, hotel_object):
        self.customer_name = customer_name
        self.hotel = hotel_object

    def generate(self):
        content = f"""
        thank yoy for your reservation!
        Her are your booking data:
        Name: {self.customer_name}
        Hotel name: {self.hotel.name}
        """
        return content


print(df)

hotel_ID = input("Enter the id of the hotel:")
hotel = Hotel(hotel_ID)

if hotel.available():
    hotel.book()
    name = input("Enter your name:")
    rv = ReservationTicket(customer_name= name, hotel_object= hotel)
    print(rv.generate())
else:
    print("Hotel is not free.")
```
