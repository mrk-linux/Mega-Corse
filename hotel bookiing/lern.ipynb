```python
import pandas  # Import pandas library for tabular data handling

# Read hotels CSV file and convert 'id' column to string type
df = pandas.read_csv("hotels.csv", dtype={"id": str})

# Hotel class for managing hotel operations
class Hotel:
    """Class to manage hotel booking, availability check, and name retrieval"""

    def __init__(self, hotel_id):
        """Store hotel ID and extract hotel name from dataframe"""
        self.hotel_id = hotel_id
        self.name = df.loc[df["id"] == self.hotel_id, "name"].squeeze()

    def book(self):
        """Change hotel availability to 'no' and save to CSV file"""
        df.loc[df["id"] == self.hotel_id, "available"] = "no"
        df.to_csv("hotels.csv", index=False)

    def available(self):
        """Check if hotel is available (returns True if available == 'yes')"""
        availability = df.loc[df["id"] == self.hotel_id, "available"].iloc[0] 
        if availability == "yes":
            return True
        else:
            return False

# ReservationTicket class for generating booking receipts
class ReservationTicket:
    """Class to generate receipt with customer name and hotel details"""

    def __init__(self, customer_name, hotel_object):
        """Store customer name and hotel object for receipt generation"""
        self.customer_name = customer_name
        self.hotel = hotel_object

    def generate(self):
        """Create receipt text including customer name and hotel name"""
        content = f"""
        thank yoy for your reservation!
        Her are your booking data:
        Name: {self.customer_name}
        Hotel name: {self.hotel.name}
        """
        return content

# Display all hotels to the user
print(df)

# Get hotel ID from user input
hotel_ID = input("Enter the id of the hotel:")

# Create hotel object based on entered ID
hotel = Hotel(hotel_ID)

# Check availability and book if possible
if hotel.available():
    hotel.book()  # Change availability to 'no'
    name = input("Enter your name:")  # Get customer name
    rv = ReservationTicket(customer_name= name, hotel_object= hotel)  # Create ticket
    print(rv.generate())  # Print the receipt
else:
    print("Hotel is not free.")  # Message when hotel is unavailable
```

## Day3

Added password authentication for credit cards using inheritance and security CSV file

```python
import pandas  # Import pandas for data manipulation

# Load hotels data and convert 'id' to string
df = pandas.read_csv("hotels.csv", dtype={"id": str})
# Load cards data as list of dictionaries for validation
df_cards = pandas.read_csv("cards.csv", dtype=str).to_dict(orient="records")
# Load security data for card password authentication
df_cards_security = pandas.read_csv("card_security.csv", dtype=str)

class Hotel:
    """Manage hotel operations: booking, availability, and name lookup"""

    def __init__(self, hotel_id):
        """Store hotel ID and fetch hotel name from dataframe"""
        self.hotel_id = hotel_id
        self.name = df.loc[df["id"] == self.hotel_id, "name"].squeeze()

    def book(self):
        """Change hotel availability to 'no' and save to CSV"""
        df.loc[df["id"] == self.hotel_id, "available"] = "no"
        df.to_csv("hotels.csv", index=False)

    def available(self):
        """Return True if hotel is available (available column == 'yes')"""
        availability = df.loc[df["id"] == self.hotel_id, "available"].iloc[0] 
        if availability == "yes":
            return True
        else:
            return False


class ReservationTicket:
    """Generate booking receipt with customer and hotel details"""

    def __init__(self, customer_name, hotel_object):
        """Store customer name and hotel object"""
        self.customer_name = customer_name
        self.hotel = hotel_object

    def generate(self):
        """Create formatted receipt text"""
        content = f"""
        thank yoy for your reservation!
        Her are your booking data:
        Name: {self.customer_name}
        Hotel name: {self.hotel.name}
        """
        return content

class Credit_Card:
    """Handle credit card validation against stored card data"""

    def __init__(self, number):
        """Store credit card number"""
        self.number = number

    def validate(self, expiration, holder, cvc):
        """Check if card details match any record in df_cards"""
        card_data = {"number": self.number, "expiration": expiration,
                    "holder": holder, "cvc": cvc}
        if card_data in df_cards:
            return True
        else:
            return False


class SecureC_C(Credit_Card):
    """Extend Credit_Card with password authentication from security CSV"""

    def authenticate(self, given_password):
        """Verify card password from security dataframe against given password"""
        try:
            password = df_cards_security.loc[df_cards_security["number"] == self.number, "password"].squeeze()
            if password == given_password:
                return True
            else:
                return False
        except:
            return False

# Display all hotels to user
print(df)

# Get hotel ID from user input
hotel_ID = input("Enter the id of the hotel:")
hotel = Hotel(hotel_ID)

# Check if hotel is available
if hotel.available():
    # Create secure credit card object with hardcoded number
    credit_card = SecureC_C(number="1234567890123456")
    # Validate card details against CSV data
    if credit_card.validate(expiration="12/26", holder="JOHN SMITH", cvc="123"):
        # Authenticate with password from security file
        if credit_card.authenticate(given_password="mypass"):
            hotel.book()  # Book the hotel
            name = input("Enter your name:")  # Get customer name
            rv = ReservationTicket(customer_name=name, hotel_object=hotel)  # Create ticket
            print(rv.generate())  # Print receipt
        else:
            print("Credit card authentication failed")
    else:
        print("There was a problem with your payment")
else:
    print("Hotel is not free.")
```
