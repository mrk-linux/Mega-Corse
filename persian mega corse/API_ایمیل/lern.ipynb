توضیحات

```python
import requests
from send_mail import send_email  # ایمپورت تابع ارسال ایمیل از فایل send_mail.py


api_key = "your api key"  # کلید API برای دسترسی به سرویس خبری
url = "The URL of the news you want to read"  # آدرس API خبری که می‌خواهید بخوانید

# ارسال درخواست به API
request = requests.get(url)

# تبدیل پاسخ دریافتی به دیکشنری (JSON)
content = request.json()

# دسترسی به عنوان و توضیحات مقالات
for article in content["articles"][:20]:  # فقط 20 مقاله اول را بررسی کن
    if article["title"] is not None:  # اگر عنوان مقاله خالی نبود
        body = "subject: Today's news" + "/n" + body + article["title"] + "/n" + article["description"] + 2 * "/n"

# تبدیل متن به فرمت UTF-8 برای ارسال ایمیل
body = body.encode("utf-8")

# ارسال ایمیل
send_email(body)
```
