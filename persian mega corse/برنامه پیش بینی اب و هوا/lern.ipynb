```python
import requests

API_key = "your API"

def get_data(place, forecast_days, kind):
    url = f"http://api.openweathermap.org/data/2.5/forecast?q={place}&appid={API_key}"
    response = requests.get(url)
    data = response.json()
    filtered_data = data["list"]
    nr_values = 8 * forecast_days
    filtered_data = filtered_data[:nr_values]
    if kind == "Temperature":
        filtered_data = [item["main"]["temp"] for item in filtered_data]

    if kind == "Sky":
        filtered_data = [item["weather"][0]["main"] for item in filtered_data]
    return filtered_data


if __name__ == "__main__":
    get_data(place="Tehran", forecast_days=2, kind="Temperature")
```

```python
import requests

# کلید API سرویس OpenWeatherMap (برای دریافت کلید در سایت https://openweathermap.org/api ثبت‌نام کنید)
API_key = "your api"

def get_data(place, forecast_days):
    """
    دریافت داده‌های پیش‌بینی هوا از API سرویس OpenWeatherMap.
    
    پارامترها:
    place (str): نام شهر (مثلاً "Tehran"، "London")
    forecast_days (int): تعداد روزهای پیش‌بینی (۱ تا ۵)
    
    خروجی:
    list: لیستی از داده‌های هواشناسی برای تعداد روزهای مشخص‌شده
          (۸ رکورد در هر روز، هر رکورد = بازه ۳ ساعته)
    """
    
    # ساخت آدرس درخواست API
    url = f"http://api.openweathermap.org/data/2.5/forecast?q={place}&appid={API_key}"
    
    # ارسال درخواست GET به API
    response = requests.get(url)
    
    # تبدیل پاسخ JSON به دیکشنری پایتون
    data = response.json()
    
    # استخراج لیست پیش‌بینی‌ها (شامل ۴۰ رکورد = ۵ روز × ۸ رکورد در روز)
    filtered_data = data["list"]
    
    # محاسبه تعداد کل رکوردهای مورد نیاز (۸ رکورد در هر روز)
    nr_values = 8 * forecast_days
    
    # محدود کردن لیست به تعداد روزهای درخواستی
    filtered_data = filtered_data[:nr_values]
    
    return filtered_data


if __name__ == "__main__":
    # تست با شهر تهران برای ۲ روز
    result = get_data(place="Tehran", forecast_days=2)
    print(result)
```
