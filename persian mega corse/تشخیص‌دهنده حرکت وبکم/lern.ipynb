# روز ۲

### تشخیص حرکت با جعبه‌های مرزی با استفاده از OpenCV

```python
import cv2              # ایمپورت کتابخانه OpenCV برای پردازش تصویر
import time             # ایمپورت کتابخانه time برای ایجاد تأخیر

video = cv2.VideoCapture(0)   # راه‌اندازی وب‌کم (۰ = دوربین پیش‌فرض)
time.sleep(1)                 # یک ثانیه مکث برای گرم شدن دوربین

first_frame = None            # متغیر برای ذخیره فریم اول به عنوان پس‌زمینه

while True:                   # شروع حلقه بی‌نهایت برای دریافت فریم‌های ویدیو
    check, frame = video.read()   # خواندن یک فریم از دوربین (پرچم موفقیت / داده فریم)
    
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    # تبدیل فریم به خاکستری برای پردازش آسان‌تر
    gray_frame_gau = cv2.GaussianBlur(gray_frame, (21, 21), 0)
    # اعمال فیلتر گاوسین برای کاهش نویز

    if first_frame is None:       # بررسی کنید که آیا این فریم اول است
        first_frame = gray_frame_gau   
        # فریم اول را به عنوان پس‌زمینه مرجع ذخیره کنید
        continue                  # بقیه حلقه را برای فریم اول رد کنید

    delta_frame = cv2.absdiff(first_frame, gray_frame_gau)  
    # پیدا کردن اختلاف مطلق بین فریم اول و فریم جاری
    thresh_frame = cv2.threshold(delta_frame, 60, 255, cv2.THRESH_BINARY)[1]  
    # اعمال آستانه‌گذاری برای برجسته کردن مناطق حرکتی
    dil_frame = cv2.dilate(thresh_frame, None, iterations=2)  
    # اتساع برای پر کردن شکاف‌های مناطق حرکتی
    
    cv2.imshow("Motion Detection", dil_frame)  
    # نمایش ماسک حرکت (سفید = حرکت تشخیص داده شده)

    contours, check = cv2.findContours(dil_frame, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)  
    # پیدا کردن کانتورهای اشیاء متحرک

    for contour in contours:      # حلقه برای تمام کانتورهای تشخیص داده شده
        if cv2.contourArea(contour) < 5000:  # رد کردن کانتورهای کوچک (نویز)
            continue              # رد کردن این کانتور و ادامه به بعدی
    
        x, y, w, h = cv2.boundingRect(contour)  
        # دریافت مختصات مستطیل محدودکننده
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 3)  
        # کشیدن مستطیل سبز دور حرکت تشخیص داده شده

    cv2.imshow("my video", frame)  # نمایش فریم اصلی با مستطیل‌های حرکت
    
    key = cv2.waitKey(1)          # یک میلی‌ثانیه منتظر ورودی از صفحه‌کلید بمان

    if key == ord("q"):           # اگر کلید فشرده‌شده 'q' باشد
        break                     # خروج از حلقه بی‌نهایت

video.release()                   # آزاد کردن دوربین و پایان دسترسی به وب‌کم
cv2.destroyAllWindows()           # بستن تمام پنجره‌های OpenCV (اختیاری اما توصیه می‌شود)
```

تمرین تشخیص حرکت را با شماره گذاری اشیا اضافه کرد

```python
import cv2              # کتابخانه پردازش تصویر
import time             # کتابخانه زمان برای تاخیر

video = cv2.VideoCapture(0)   # راه‌اندازی وب‌کم
time.sleep(1)                 # تاخیر برای گرم شدن دوربین

first_frame = None            # ذخیره فریم اول به عنوان پس‌زمینه

while True:
    check, frame = video.read()   # خواندن فریم از دوربین
    
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)  
    # تبدیل به خاکستری
    gray_blur = cv2.GaussianBlur(gray, (21, 21), 0)  
    # اعمال بلور برای کاهش نویز

    if first_frame is None:        # اگر این فریم اول است
        first_frame = gray_blur    # ذخیره به عنوان پس‌زمینه
        continue                   # رفتن به فریم بعدی

    delta = cv2.absdiff(first_frame, gray_blur)  
    # پیدا کردن اختلاف از پس‌زمینه (تشخیص حرکت)
    thresh = cv2.threshold(delta, 50, 255, cv2.THRESH_BINARY)[1]  
    # اعمال آستانه‌گذاری
    dilate = cv2.dilate(thresh, None, iterations=2)  
    # پر کردن شکاف‌های مناطق حرکتی

    contours, _ = cv2.findContours(dilate, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)  
    # پیدا کردن اشیاء

    object_number = 1   # شمارنده اشیاء
    
    for contour in contours:   # برای هر شیء تشخیص داده شده
        if cv2.contourArea(contour) < 3000:  # اگر مساحت خیلی کوچک است
            continue            # رد کردن (نویز است)
        
        x, y, w, h = cv2.boundingRect(contour)  
        # دریافت مختصات مستطیل محدودکننده
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)  
        # کشیدن مستطیل سبز
        

        cv2.putText(frame, f"Object {object_number}", (x, y-10),  
                    # نمایش شماره شیء
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
        
        object_number += 1   # افزایش شمارنده

    cv2.imshow("Motion Detection", frame)  # نمایش ویدیو

    if cv2.waitKey(1) == ord("q"):  # اگر کلید 'q' فشرده شود
        break                       # خروج از حلقه

video.release()        # آزادسازی دوربین
cv2.destroyAllWindows()  # بستن پنجره‌ها
```

```python
import cv2
import time
import glob  # برای کار با فایل‌های تصویری و لیست کردن آنها
from Email_ing import send_email


video = cv2.VideoCapture(0)
time.sleep(1)

first_frame = None
status_list = []
count = 1  # شمارنده برای نام‌گذاری تصاویر ذخیره شده

while True:
    status = 0
    check, frame = video.read()

    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray_frame_gau = cv2.GaussianBlur(gray_frame, (21, 21), 0)

    if first_frame is None:
        first_frame = gray_frame_gau

    delta_frame = cv2.absdiff(first_frame, gray_frame_gau)
    thresh_frame = cv2.threshold(delta_frame, 60, 255, cv2.THRESH_BINARY)[1]
    dil_frame = cv2.dilate(thresh_frame, None, iterations=2)
    cv2.imshow("My Video", dil_frame)

    contours, check = cv2.findContours(dil_frame, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    for contour in contours:
        if cv2.contourArea(contour) < 3000:
            continue
        
        x, y, w, h = cv2.boundingRect(contour)  
        rectangle = cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)  

        if rectangle.any():
            status = 1
            cv2.imwrite(f"images/{count}.png", frame)  # ذخیره تصویر در پوشه images
            count = count + 1  # افزایش شمارنده
            all_images = glob.glob("images/*.png")  # دریافت لیست تمام تصاویر ذخیره شده
            index = int(len(all_images) / 2)  # انتخاب تصویر وسط
            image_with_object = all_images[index]  # تصویر وسط

    status_list.append(status)
    status_list = status_list[-2:]

    if status_list[0] == 1 and status_list[1] == 0:
        send_email(image_with_object)

        cv2.putText(frame, f"Object {object_number}", (x, y-10),  
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
        
        object_number += 1

    cv2.imshow("Motion Detection", frame)

    if cv2.waitKey(1) == ord("q"):
        break

video.release()
cv2.destroyAllWindows()
```

### ارسال ایمیل

```python
import smtplib  # برای اتصال به سرور ایمیل و ارسال پیام‌ها
import imghdr  # برای تشخیص نوع تصویر (مثلاً png، jpg)
from email.message import EmailMessage  # برای ایجاد ایمیل با پیوست‌ها

# اطلاعات ورود به جیمیل
PASSWORD = "your_password_here"  # رمز برنامه جیمیل (نه رمز معمولی شما)
SENDER = "app8flask@gmail.com"  # آدرس ایمیل فرستنده
RECEIVER = "app8flask@gmail.com"  # آدرس ایمیل گیرنده

def send_email(image_path):  # تابع برای ارسال ایمیل با تصویر
    email_message = EmailMessage()  # ایجاد یک پیام ایمیل جدید
    email_message["Subject"] = "New customer showed up!"  # موضوع ایمیل
    email_message.set_content("Hey, we just saw a new customer!")  # متن بدنه ایمیل

    with open(image_path, "rb") as file:  # باز کردن فایل تصویر در حالت باینری
        content = file.read()  # خواندن محتوای تصویر
        # اضافه کردن تصویر به عنوان پیوست
        email_message.add_attachment(
            content,  # داده تصویر
            maintype="image",  # نوع اصلی = تصویر
            subtype=imghdr.what(None, content),  # تشخیص نوع تصویر (png، jpg و غیره)
            filename=image_path.split("/")[-1]  # نام فایل (بدون مسیر)
        )

    # اتصال به سرور جیمیل
    gmail = smtplib.SMTP("smtp.gmail.com", 587)  # سرور و پورت جیمیل
    gmail.ehlo()  # سلام به سرور
    gmail.starttls()  # فعال کردن امنیت (TLS)
    gmail.login(SENDER, PASSWORD)  # ورود به حساب جیمیل
    gmail.sendmail(SENDER, RECEIVER, email_message.as_string())  # ارسال ایمیل
    gmail.quit()  # قطع ارتباط
    print("✅ Email sent successfully!")  # پیام موفقیت

if __name__ == "__main__":  # اگر این فایل مستقیماً اجرا شود
    send_email(image_path="images/19.png")  # ارسال ایمیل با تصویر مشخص شده
```
