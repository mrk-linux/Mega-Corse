```python
import pandas  # وارد کردن کتابخانه pandas برای مدیریت داده‌های جدولی

# خواندن فایل CSV هتل‌ها و تبدیل ستون 'id' به نوع رشته
df = pandas.read_csv("hotels.csv", dtype={"id": str})

# کلاس هتل برای مدیریت عملیات مربوط به هتل
class Hotel:
    """کلاس برای مدیریت رزرو هتل، بررسی موجودی و دریافت نام هتل"""

    def __init__(self, hotel_id):
        """ذخیره شناسه هتل و استخراج نام هتل از دیتافریم"""
        self.hotel_id = hotel_id
        self.name = df.loc[df["id"] == self.hotel_id, "name"].squeeze()

    def book(self):
        """تغییر وضعیت موجودی هتل به 'no' و ذخیره در فایل CSV"""
        df.loc[df["id"] == self.hotel_id, "available"] = "no"
        df.to_csv("hotels.csv", index=False)

    def available(self):
        """بررسی موجود بودن هتل (در صورت موجود بودن مقدار 'yes' برمی‌گرداند)"""
        availability = df.loc[df["id"] == self.hotel_id, "available"].iloc[0] 
        if availability == "yes":
            return True
        else:
            return False

# کلاس بلیط رزرو برای تولید رسیدهای رزرو
class ReservationTicket:
    """کلاس برای تولید رسید با جزئیات نام مشتری و هتل"""

    def __init__(self, customer_name, hotel_object):
        """ذخیره نام مشتری و شیء هتل برای تولید رسید"""
        self.customer_name = customer_name
        self.hotel = hotel_object

    def generate(self):
        """ایجاد متن رسید شامل نام مشتری و نام هتل"""
        content = f"""
        با تشکر از رزرو شما!
        اطلاعات رزرو شما به شرح زیر است:
        نام: {self.customer_name}
        نام هتل: {self.hotel.name}
        """
        return content

# نمایش همه هتل‌ها به کاربر
print(df)

# دریافت شناسه هتل از ورودی کاربر
hotel_ID = input("شناسه هتل را وارد کنید:")

# ایجاد شیء هتل بر اساس شناسه وارد شده
hotel = Hotel(hotel_ID)

# بررسی موجودی و رزرو در صورت امکان
if hotel.available():
    hotel.book()  # تغییر وضعیت موجودی به 'no'
    name = input("نام خود را وارد کنید:")  # دریافت نام مشتری
    rv = ReservationTicket(customer_name=name, hotel_object=hotel)  # ایجاد بلیط
    print(rv.generate())  # چاپ رسید
else:
    print("هتل موجود نیست.")  # پیام زمانی که هتل در دسترس نیست
```

```python
import pandas  # وارد کردن pandas برای دستکاری داده‌ها

# بارگذاری داده‌های هتل‌ها و تبدیل 'id' به رشته
df = pandas.read_csv("hotels.csv", dtype={"id": str})
# بارگذاری داده‌های کارت‌ها به صورت لیستی از دیکشنری‌ها (هر سطر = یک دیکشنری)
df_cards = pandas.read_csv("cards.csv", dtype=str).to_dict(orient="records")

class Hotel:
    """مدیریت عملیات هتل: رزرو، بررسی موجودی، و جستجوی نام"""

    def __init__(self, hotel_id):
        """ذخیره شناسه هتل و دریافت نام هتل از دیتافریم"""
        self.hotel_id = hotel_id
        self.name = df.loc[df["id"] == self.hotel_id, "name"].squeeze()

    def book(self):
        """تغییر وضعیت موجودی هتل به 'no' و ذخیره در فایل CSV"""
        df.loc[df["id"] == self.hotel_id, "available"] = "no"
        df.to_csv("hotels.csv", index=False)

    def available(self):
        """در صورت موجود بودن هتل (ستون available == 'yes') مقدار True برمی‌گرداند"""
        availability = df.loc[df["id"] == self.hotel_id, "available"].iloc[0] 
        if availability == "yes":
            return True
        else:
            return False


class ReservationTicket:
    """تولید رسید رزرو با جزئیات مشتری و هتل"""

    def __init__(self, customer_name, hotel_object):
        """ذخیره نام مشتری و شیء هتل"""
        self.customer_name = customer_name
        self.hotel = hotel_object

    def generate(self):
        """ایجاد متن رسید با قالب‌بندی مناسب"""
        content = f"""
        با تشکر از رزرو شما!
        اطلاعات رزرو شما به شرح زیر است:
        نام: {self.customer_name}
        نام هتل: {self.hotel.name}
        """
        return content

class Credit_Card:
    """مدیریت اعتبارسنجی کارت اعتباری بر اساس داده‌های ذخیره شده"""

    def __init__(self, number):
        """ذخیره شماره کارت اعتباری"""
        self.number = number

    def validate(self, expiration, holder, cvc):
        """بررسی تطابق جزئیات کارت با هر رکورد در df_cards"""
        card_data = {"number": self.number, "expiration": expiration,
                    "holder": holder, "cvc": cvc}
        if card_data in df_cards:
            return True
        else:
            return False


# نمایش همه هتل‌ها به کاربر
print(df)

# دریافت شناسه هتل از ورودی کاربر
hotel_ID = input("شناسه هتل را وارد کنید:")
hotel = Hotel(hotel_ID)

# بررسی موجود بودن هتل
if hotel.available():
    # ایجاد شیء کارت اعتباری با شماره ثابت (کد سخت)
    credit_card = Credit_Card(number="1234567890123456")
    # اعتبارسنجی جزئیات کارت بر اساس داده‌های CSV
    if credit_card.validate(expiration="12/26", holder="JOHN SMITH", cvc="123"):
        hotel.book()  # رزرو هتل
        name = input("نام خود را وارد کنید:")  # دریافت نام مشتری
        rv = ReservationTicket(customer_name=name, hotel_object=hotel)  # ایجاد بلیط
        print(rv.generate())  # چاپ رسید
    else:
        print("مشکلی در پرداخت شما وجود دارد")
else:
    print("هتل موجود نیست.")
```

## روز سوم

اضافه شدن احراز هویت رمز عبور برای کارت های اعتباری با استفاده از فایل CSV وراثت و امنیتی

```python
import pandas  # وارد کردن pandas برای دستکاری داده‌ها

# بارگذاری داده‌های هتل‌ها و تبدیل 'id' به رشته
df = pandas.read_csv("hotels.csv", dtype={"id": str})
# بارگذاری داده‌های کارت‌ها به صورت لیستی از دیکشنری‌ها برای اعتبارسنجی
df_cards = pandas.read_csv("cards.csv", dtype=str).to_dict(orient="records")
# بارگذاری داده‌های امنیتی برای احراز هویت رمز عبور کارت
df_cards_security = pandas.read_csv("card_security.csv", dtype=str)

class Hotel:
    """مدیریت عملیات هتل: رزرو، بررسی موجودی، و جستجوی نام"""

    def __init__(self, hotel_id):
        """ذخیره شناسه هتل و دریافت نام هتل از دیتافریم"""
        self.hotel_id = hotel_id
        self.name = df.loc[df["id"] == self.hotel_id, "name"].squeeze()

    def book(self):
        """تغییر وضعیت موجودی هتل به 'no' و ذخیره در فایل CSV"""
        df.loc[df["id"] == self.hotel_id, "available"] = "no"
        df.to_csv("hotels.csv", index=False)

    def available(self):
        """در صورت موجود بودن هتل (ستون available == 'yes') مقدار True برمی‌گرداند"""
        availability = df.loc[df["id"] == self.hotel_id, "available"].iloc[0] 
        if availability == "yes":
            return True
        else:
            return False


class ReservationTicket:
    """تولید رسید رزرو با جزئیات مشتری و هتل"""

    def __init__(self, customer_name, hotel_object):
        """ذخیره نام مشتری و شیء هتل"""
        self.customer_name = customer_name
        self.hotel = hotel_object

    def generate(self):
        """ایجاد متن رسید با قالب‌بندی مناسب"""
        content = f"""
        با تشکر از رزرو شما!
        اطلاعات رزرو شما به شرح زیر است:
        نام: {self.customer_name}
        نام هتل: {self.hotel.name}
        """
        return content

class Credit_Card:
    """مدیریت اعتبارسنجی کارت اعتباری بر اساس داده‌های ذخیره شده"""

    def __init__(self, number):
        """ذخیره شماره کارت اعتباری"""
        self.number = number

    def validate(self, expiration, holder, cvc):
        """بررسی تطابق جزئیات کارت با هر رکورد در df_cards"""
        card_data = {"number": self.number, "expiration": expiration,
                    "holder": holder, "cvc": cvc}
        if card_data in df_cards:
            return True
        else:
            return False


class SecureC_C(Credit_Card):
    """تعمیم کلاس Credit_Card با احراز هویت رمز عبور از فایل CSV امنیتی"""

    def authenticate(self, given_password):
        """بررسی رمز عبور کارت از دیتافریم امنیتی در برابر رمز عبور داده شده"""
        try:
            password = df_cards_security.loc[df_cards_security["number"] == self.number, "password"].squeeze()
            if password == given_password:
                return True
            else:
                return False
        except:
            return False

# نمایش همه هتل‌ها به کاربر
print(df)

# دریافت شناسه هتل از ورودی کاربر
hotel_ID = input("شناسه هتل را وارد کنید:")
hotel = Hotel(hotel_ID)

# بررسی موجود بودن هتل
if hotel.available():
    # ایجاد شیء کارت اعتباری امن با شماره ثابت (کد سخت)
    credit_card = SecureC_C(number="1234567890123456")
    # اعتبارسنجی جزئیات کارت بر اساس داده‌های CSV
    if credit_card.validate(expiration="12/26", holder="JOHN SMITH", cvc="123"):
        # احراز هویت با رمز عبور از فایل امنیتی
        if credit_card.authenticate(given_password="mypass"):
            hotel.book()  # رزرو هتل
            name = input("نام خود را وارد کنید:")  # دریافت نام مشتری
            rv = ReservationTicket(customer_name=name, hotel_object=hotel)  # ایجاد بلیط
            print(rv.generate())  # چاپ رسید
        else:
            print("احراز هویت کارت اعتباری ناموفق بود")
    else:
        print("مشکلی در پرداخت شما وجود دارد")
else:
    print("هتل موجود نیست.")
```
