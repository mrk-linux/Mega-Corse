```python
import pandas as pd
import glob
from fpdf import FPDF
from pathlib import Path

filepaths = glob.glob("Invoices/*.xlsx")

for filepath in filepaths:
    df = pd.read_excel(filepath)
    pdf = FPDF(orientation="P", unit="mm", format="A4")
    pdf.add_page()
    
    filename = Path(filepath).stem 
    invoice_nr = filename.split("-")[0]
    
    pdf.set_font(family="Times", size=16, style="B")
    pdf.cell(w=50, h=8, txt=f"Invoice nr.{invoice_nr}", ln=1)
    
    pdf.output(f"pdfs/{filename}.pdf")
```

```python
import pandas as pd
import glob
from fpdf import FPDF
from pathlib import Path


filepaths = glob.glob("Invoices/*.xlsx")

for filepath in filepaths:
    df = pd.read_excel(filepath)

    pdf = FPDF(orientation="P", unit="mm", format="A4")
    pdf.add_page()
    
    filename = Path(filepath).stem 
    invoice_nr, date = filename.split("-")
    
    pdf.set_font(family="Times", size=16, style="B")
    pdf.cell(w=50, h=8, txt=f"Invoice nr.{invoice_nr}", ln=1)
    
    pdf.set_font(family="Times", size=16, style="B")
    pdf.cell(w=50, h=8, txt=f"Date.{date}", ln= 1)

    #a header
    columns = df.columns
    columns = [item.replace("_", " ").title() for item in columns]
    pdf.set_font(family="Times", size= 10, style="B")
    pdf.set_text_color(80, 80 ,80 )
    pdf.cell(w= 30, h= 8, txt= columns[0], border= 1)
    pdf.cell(w= 70, h= 8, txt= columns[1], border= 1)
    pdf.cell(w= 35, h= 8, txt= columns[2], border= 1)
    pdf.cell(w= 30, h= 8, txt= columns[3], border= 1)
    pdf.cell(w= 30, h= 8, txt= columns[4], border= 1, ln= 1)

    #add raws to the table
    for index, row in df.iterrows():
        pdf.set_font(family="Times", size= 10)
        pdf.set_text_color(80, 80 ,80 )
        pdf.cell(w= 30, h= 8, txt= str(row["product_id"]), border= 1)
        pdf.cell(w= 70, h= 8, txt= str(row["product_name"]), border= 1)
        pdf.cell(w= 35, h= 8, txt= str(row["amount_purchased"]), border= 1)
        pdf.cell(w= 30, h= 8, txt= str(row["price_per_unit"]), border= 1)
        pdf.cell(w= 30, h= 8, txt= str(row["total_price"]), border= 1, ln= 1)

    pdf.output(f"Pdfs/{filename}.pdf")
```

توضیحات کد

```python
import pandas as pd
import glob
from fpdf import FPDF
from pathlib import Path


filepaths = glob.glob("Invoices/*.xlsx")  # دریافت همه فایل‌های اکسل از پوشه Invoices

for filepath in filepaths:
    df = pd.read_excel(filepath)  # خواندن فایل اکسل در دیتافریم

    pdf = FPDF(orientation="P", unit="mm", format="A4")  # ایجاد شیء PDF (عمودی، میلی‌متر، A4)
    pdf.add_page()  # افزودن صفحه جدید
    
    filename = Path(filepath).stem  # دریافت نام فایل بدون پسوند
    invoice_nr, date = filename.split("-")  # استخراج شماره فاکتور و تاریخ از نام فایل
    
    pdf.set_font(family="Times", size=16, style="B")  # تنظیم فونت (تایمز، بولت، 16pt)
    pdf.cell(w=50, h=8, txt=f"Invoice nr.{invoice_nr}", ln=1)  # نوشتن شماره فاکتور
    
    pdf.set_font(family="Times", size=16, style="B")  # تنظیم مجدد فونت
    pdf.cell(w=50, h=8, txt=f"Date.{date}", ln=1)  # نوشتن تاریخ

    columns = df.columns  # دریافت نام ستون‌ها از دیتافریم
    columns = [item.replace("_", " ").title() for item in columns]  # تمیز کردن نام ستون‌ها: جایگزینی زیرخط با فاصله و بزرگ کردن حروف
    
    pdf.set_font(family="Times", size=10, style="B")  # تنظیم فونت برای هدر (بولت، 10pt)
    pdf.set_text_color(80, 80, 80)  # تنظیم رنگ متن به خاکستری
    pdf.cell(w=30, h=8, txt=columns[0], border=1)  # ستون 1: product_id
    pdf.cell(w=70, h=8, txt=columns[1], border=1)  # ستون 2: product_name
    pdf.cell(w=30, h=8, txt=columns[2], border=1)  # ستون 3: amount_purchased
    pdf.cell(w=30, h=8, txt=columns[3], border=1)  # ستون 4: price_per_unit
    pdf.cell(w=30, h=8, txt=columns[4], border=1, ln=1)  # ستون 5: total_price

    for index, row in df.iterrows():  # پیمایش هر ردیف در دیتافریم
        pdf.set_font(family="Times", size=10)  # تنظیم فونت برای ردیف‌های داده (معمولی، 10pt)
        pdf.set_text_color(80, 80, 80)  # حفظ رنگ خاکستری
        pdf.cell(w=30, h=8, txt=str(row["product_id"]), border=1)  # نوشتن product_id
        pdf.cell(w=70, h=8, txt=str(row["product_name"]), border=1)  # نوشتن product_name
        pdf.cell(w=30, h=8, txt=str(row["amount_purchased"]), border=1)  # نوشتن amount_purchased
        pdf.cell(w=30, h=8, txt=str(row["price_per_unit"]), border=1)  # نوشتن price_per_unit
        pdf.cell(w=30, h=8, txt=str(row["total_price"]), border=1, ln=1)  # نوشتن total_price

    pdf.output(f"pdfs/{filename}.pdf")  # ذخیره PDF در پوشه Pdfs
```
